```python
import pygmt
import matplotlib.pyplot as plt
import numpy as np


region = 'd'
spacing = '1d'

res = pygmt.blockmedian('output/tracers_0040.csv', region=region, incols='1,2,0', spacing=spacing, output_type='numpy')

resgrd = pygmt.sphinterpolate(res, region=region, spacing=spacing)

plt.contourf(resgrd, vmin=0, vmax=50, cmap='magma_r', levels=np.arange(0,100,5))
plt.colorbar()
```
