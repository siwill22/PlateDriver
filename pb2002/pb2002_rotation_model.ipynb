```python
import pandas as pd
import math
import pygplates


rotations = pd.read_fwf('/Users/Simon/Data/GIS/BirdPlateBoundaries/PB2002_poles.dat',
                        header=None,
                        colspecs=[(0,2),(2,12),(12,22),(22,31),(31,-1)],
                        names=('PID','PLat','PLong','PAngle','Comment'),
                        keep_default_na=False)

plate_list = rotations.PID.unique()

rotation_features = []

for index,plate in enumerate(plate_list):
    subset=rotations.where(rotations.PID.str.match(plate)).dropna()
    
    rotation_feature = pygplates.Feature(pygplates.FeatureType.create_gpml('TotalReconstructionSequence'))
    
    gpml_irregular_sampling = pygplates.GpmlIrregularSampling(
                [pygplates.GpmlTimeSample(
                        pygplates.GpmlFiniteRotation(
                                pygplates.FiniteRotation(
                                        pygplates.PointOnSphere(1,0,0),0)),
                        0,plate),
                 pygplates.GpmlTimeSample(
                        pygplates.GpmlFiniteRotation(
                                pygplates.FiniteRotation(
                                        pygplates.PointOnSphere(subset.PLat.iloc[0],subset.PLong.iloc[0]),-1*math.radians(subset.PAngle.iloc[0]))),
                        1,'%s, %s' % (plate,subset.Comment.iloc[0]))
                ])
    
    rotation_feature = pygplates.Feature.create_total_reconstruction_sequence(
        0,
        index+101,
        total_reconstruction_pole=gpml_irregular_sampling,
        name=['pb2002: %s' % plate])

    rotation_features.append(rotation_feature)

#print rotation_features

rotation_file = pygplates.FeatureCollection(rotation_features)
feature_collection_file_format_registry = pygplates.FeatureCollectionFileFormatRegistry()
feature_collection_file_format_registry.write(rotation_file,'pb2002_gplates2.rot')

```
